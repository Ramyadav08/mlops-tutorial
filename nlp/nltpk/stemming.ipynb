```python
#stemming
words=["play", "playing", "played", "plays", "player","going", "goes", "gone", "go"]
```

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
for word in words:
    print(stemming.stem(word))
```

```python
from nltk.stem import  RegexpStemmer
stemming=RegexpStemmer('ing$')
for word in words:
    print(stemming.stem(word))
```

```python
#snowballstmer
from nltk.stem import SnowballStemmer
stemming=SnowballStemmer('english')
for word in words:
    print(stemming.stem(word))
```

```python
#snawballstmer is more aggressive than porter stemmer
```
