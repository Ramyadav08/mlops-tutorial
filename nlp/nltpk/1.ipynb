```python
corpus='''hello welcome to the world of python programming.
python is a great language for data science! and machine learning.
the python community is very supportive! and helpful.
learning python can open up many career opportunities in the tech industry.'''
```

```python
print(corpus)
```

```python
#tokenize the corpus into sentences
from nltk.tokenize import sent_tokenize
documents=sent_tokenize(corpus)
```

```python
print(documents)
```

```python
type(documents)
```

```python
#word tokenization
from nltk.tokenize import word_tokenize
tokens=word_tokenize(corpus)
```

```python
print(tokens)
```

```python
from nltk.tokenize import treebank
tokens=treebank.TreebankWordTokenizer().tokenize(corpus)
```

```python
print(tokens)
```
