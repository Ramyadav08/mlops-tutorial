```python
import pandas as pd   
import numpy as np         
import seaborn as sns    
import matplotlib.pyplot as plt               
%matplotlib inline
```

```python
#feature engineering
df["Date"]=df["Date_of_Journey"].str.split('/').str[0]
df["Month"]=df["Date_of_Journey"].str.split('/').str[1]
df["Year"]=df["Date_of_Journey"].str.split('/').str[2]
```

```python
df.head()
```

```python
df.info()
```

```python
df["Date"]=df['Date'].astype(int)
df["Month"]=df['Month'].astype(int)
df["Year"]=df['Year'].astype(int)
```

```python
df.head()
```

```python
df.info()
```

```python
#drop the Date_of_Journey
df.drop('Date_of_Journey',axis=1,inplace=True)
df.head()
```

```python
df['Arrival_Time']=df['Arrival_Time'].apply(lambda x:x.split(' ')[0])
```

```python
df['Arrival_Hour']=df['Arrival_Time'].str.split(":").str[0]
df['Arrival_Minute']=df['Arrival_Time'].str.split(":").str[1]
df.head()
```

```python
df.info()
```

```python
df['Arrival_Hour']=df['Arrival_Hour'].astype(int)
df['Arrival_Minute']=df['Arrival_Minute'].astype(int)
df.head()
```

```python
#drop the Arrival_Time 
df.drop('Arrival_Time',axis=1,inplace=True)
df.head()
```

```python
df.info()
```

```python
df['Dep_Hour']=df['Dep_Time'].str.split(":").str[0]
df['Dep_Minute']=df['Dep_Time'].str.split(":").str[1]
df.head()
```

```python
df['Dep_Hour']=df['Dep_Hour'].astype(int)
df['Dep_Minute']=df['Dep_Minute'].astype(int)
df.head()
```

```python
df.drop('Dep_Time',axis=1, inplace=True)
df.head(2)
```

```python
df.info()
```

```python
df["Total_Stops"].unique()
```

```python
df[df["Total_Stops"].isnull()]
```

```python
df['Total_Stops'].mode()[0]
```

```python
df["Total_Stops"]=df["Total_Stops"].map({'non-stop':0,'1 stop':1,'2 stops':2,'3 stops':3,'4 stops':4,np.nan:1})
```

```python
df.head(2)
```

```python
df.info()
```

```python
df.drop('Route',axis=1,inplace=True)
df.head(2)
```

```python
df['Duration'].str.split(' ').str[0].str.split('h').str[0]
```

```python
df['Duration'].str.split(' ').str[1].str.split('m').str[0].fillna(0)
```

```python
# df['Duration_hour']=df['Duration'].str.split(' ').str[0].str.split('h').str[0]
# df['Duration_minute']=df['Duration'].str.split(' ').str[1].str.split('m').str[0].fillna(0)
df['Duration_hour'] = (
    df['Duration']
    .str.extract(r'(\d+)h')[0]   
    .fillna('0')
    .astype(int)
)
df['Duration_minute'] = (
    df['Duration']
    .str.extract(r'(\d+)m')[0]
    .fillna('0')
    .astype(int)
)
df.head(2)
```

```python
from sklearn.preprocessing import OneHotEncoder
encoder=OneHotEncoder()
```

```python
encoder.fit_transform(df[['Airline','Source','Destination','Additional_Info']]).toarray()
```

```python
encoded_df=pd.DataFrame(encoder.fit_transform(df[['Airline','Source','Destination','Additional_Info']]).toarray(),columns=encoder.get_feature_names_out())
```

```python
encoded_df = encoded_df.reset_index(drop=True)
df = df.reset_index(drop=True)
```

```python
df_final = pd.concat([df, encoded_df], axis=1)
```

```python
df_final = df_final.drop(['Airline','Source','Destination','Additional_Info'], axis=1)
```
